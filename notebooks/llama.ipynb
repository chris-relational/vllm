# `meta-llama/Llama-3.2-1B-Instruct`

```python
llm = util.ModelConfig.from_pretrained("meta-llama/Llama-3.2-1B-Instruct")
```

```python
# Call generate using a plain and an annotated prompt
plain_prompt = "Write 5 sentences of at least 5 words each, so that each sentence contains at least 1 word more than its orevious sentence."

messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": plain_prompt}
]

# tokenize=False will make the tokenizer return a string, otherwise a list with the token_ids is returned
annotated_prompt = llm.tokenizer.apply_chat_template(messages, tokenize=False)
inputs = llm.tokenizer(text=annotated_prompt, return_tensors="pt")
```

## Batched generation

```python
print(*llm.generate(**inputs, max_new_tokens=256))
```

## Streamed Generation

```python
from transformers import TextStreamer
_ = llm.model.generate(**inputs, streamer=TextStreamer(llm.tokenizer), max_new_tokens=256)
```

# `meta-llama/Llama-3.2-1B`

```python
llm = util.ModelConfig.from_pretrained("meta-llama/Llama-3.2-1B")
```

```python
from transformers import AutoTokenizer

# Call generate using a plain and an annotated prompt
plain_prompt = "Write 5 sentences of at least 5 words each, so that each sentence contains at least 1 word more than its orevious sentence."

messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": plain_prompt}
]

# this model is old, without a chat template. We exploit another tokenizer to convert the chat template to a string
chat_tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-1B-Instruct")

# tokenize=False will make the tokenizer return a string, otherwise a list with the token_ids is returned
annotated_prompt = chat_tokenizer.apply_chat_template(messages, tokenize=False)
inputs = llm.tokenizer(text=annotated_prompt, return_tensors="pt")
```

## Batched generation

```python
# print(*llm.generate(**inputs, max_new_tokens=256))
print(llm.tokenizer.batch_decode(llm.model.generate(**inputs, max_new_tokens=256)))
```

## Streamed Generation

```python
from transformers import TextStreamer
_ = llm.model.generate(**inputs, streamer=TextStreamer(llm.tokenizer), max_new_tokens=256)
```
