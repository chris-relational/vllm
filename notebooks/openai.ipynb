# TL; DR

The cells here contain inference invocations of HF LLMs with the OpenAI API,  

using curl, the Python HTTP requests API or the OpenAI Python SDK (built on top of the REST API).  

For inference requests using Huggingface-Hub's InferenceClient see `huggingface-hub.ipynb`

```python
%%bash
curl http://localhost:8000/v1/openai/chat/completions \
    -H "Content-Type: application/json" \
    -d '{
        "model": "masint/tiny-random-llama",
        "messages": [            
            {"role": "user", "content": "Who won the world series in 2020?"}
        ]
    }'
```

```python
%%bash
curl -X POST "http://localhost:8000/v1/chat/completions" \
	-H "Content-Type: application/json" \
	--data '{
		"model": "mistralai/Mistral-7B-Instruct-v0.1",
		"messages": [
			{
				"role": "user",
				"content": "What is the capital of France?"
			}
		]
	}'
```
